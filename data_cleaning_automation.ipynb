```python
# ==========================================
# DATA CLEANING & REPORTING AUTOMATION
# ==========================================

# Import Libraries
import pandas as pd

# ==========================================
# LOAD DATASET
# ==========================================

df = pd.read_csv(
    '/content/Sample - Superstore.csv',
    encoding='cp1252'
)

print("========== ORIGINAL DATA ==========")
print(df.head())

# ==========================================
# CHECK MISSING VALUES
# ==========================================

print("\n========== MISSING VALUES ==========")
print(df.isnull().sum())

# Fill missing numeric values with mean
numeric_columns = df.select_dtypes(include=['float64', 'int64']).columns

for col in numeric_columns:
    df[col].fillna(df[col].mean(), inplace=True)

# Fill missing categorical values with mode
categorical_columns = df.select_dtypes(include=['object']).columns

for col in categorical_columns:
    df[col].fillna(df[col].mode()[0], inplace=True)

# ==========================================
# REMOVE DUPLICATES
# ==========================================

duplicates_before = df.duplicated().sum()

df.drop_duplicates(inplace=True)

duplicates_after = df.duplicated().sum()

print("\n========== DUPLICATE RECORDS ==========")
print(f"Duplicates Before Removal: {duplicates_before}")
print(f"Duplicates After Removal : {duplicates_after}")

# ==========================================
# GENERATE SUMMARY REPORT
# ==========================================

summary_report = pd.DataFrame({
    'Metric': [
        'Total Rows',
        'Total Columns',
        'Missing Values Remaining',
        'Duplicate Records Remaining'
    ],
    'Value': [
        df.shape[0],
        df.shape[1],
        df.isnull().sum().sum(),
        df.duplicated().sum()
    ]
})

print("\n========== SUMMARY REPORT ==========")
print(summary_report)

# ==========================================
# EXPORT CLEANED DATA
# ==========================================

df.to_excel(
    'cleaned_superstore_data.xlsx',
    index=False
)

# Export summary report
summary_report.to_excel(
    'summary_report.xlsx',
    index=False
)

print("\n========== EXPORT COMPLETED ==========")
print("Cleaned dataset saved as: cleaned_superstore_data.xlsx")
print("Summary report saved as: summary_report.xlsx")

print("\nData Cleaning & Reporting Automation Completed Successfully!")
```
